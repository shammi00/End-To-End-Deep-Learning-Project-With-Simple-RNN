```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[
'the glass of juice',
'the cup of tea',
'I am a good boy',
'I am a good developer',
"understand the meaning of words",
'your videos are good', ]
```

```python
sent
```

```python
## define the vocabulary size
voc_size=10000
```

```python
## one hot representation of the sentences
one_hot_repr=[one_hot(words, voc_size) for words in sent]
one_hot_repr
```

```python
# word embedding representation
from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_len=max([len(i.split()) for i in sent])
embeded_docs=pad_sequences(one_hot_repr, padding='pre', maxlen=sent_len)
embeded_docs
```

```python
print(embeded_docs)
```

```python
#feature representation (vector size)
dim=10
```

```python
# Create the embedding layer
embedding_layer=Embedding(voc_size, dim, input_length=sent_len)
```

```python
embeded_docs[0]
```

```python
# initial weight representation before training
embedding_layer(embeded_docs)
```
